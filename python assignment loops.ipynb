```python
#even numbers (1 to 20)
for i in range(1, 21):
    if i % 2 == 0:
        print(i)
```

```python
# odd numbers (1 to 20)
for i in range(1, 21):
    if i % 2 != 0:
        print(i)
```

```python
# five found check
for i in range(1, 11):
    if i == 5:
        print('Five found!')
    else:
        print(i)
```

```python
# numbers greater then 5
numbers = [3, 8, 12, 1, 9]
for num in numbers:
    if num > 5:
        print(num)
```

```python
#divisible by 3
for i in range(1, 16):
    if i % 3 == 0:
        print(i)
```

```python
#positive negative or zero
numbers = [2, -5, 10, -8, 0]
for num in numbers:
    if num > 0:
        print('Positive')
    elif num < 0:
        print('Negative')
    else:
        print('Zero')
```

```python
#break loop at 7
for i in range(1, 11):
    if i == 7:
        break
    print(i)
```

```python
#continue (skip 4)
for i in range(1, 11):
    if i == 4:
        continue
    print(i)
```

```python
#continue skip (4)
text = 'python'
vowels = 'aeiou'
for char in text:
    if char in vowels:
        print(char)
```

```python
#Adult or minor check
ages = [12, 18, 25, 15]
for age in ages:
    if age >= 18:
        print('Adult')
    else:
        print('Minor')
```

```python
#search name sara
names = ['Ali', 'Sara', 'Zain']
for name in names:
    if name == 'Sara':
        print('Sara is present!')
```

```python
#fizbux divisible by 3 and 5
for i in range(1, 21):
    if i % 3 == 0 and i % 5 == 0:
        print('FizzBuzz')
    else:
        print(i)
```

```python
#double even number
for i in range(1, 11):
    if i % 2 == 0:
        print(i * 2)
    else:
        print(i)
```

```python
#names starting with A
names = ['Ali', 'Bilal', 'Ahmed', 'Sana']
for name in names:
    if name.startswith('A'):
        print(name)
```

```python
#count apple occurences
fruits = ['apple', 'mango', 'apple', 'banana', 'apple']
count = 0
for fruit in fruits:
    if fruit == 'apple':
        count += 1
print('Apple count:', count)
```

```python
#pass or fail
marks_list = [45, 80, 30, 95, 40]
for marks in marks_list:
    if marks > 50:
        print('Pass')
    else:
        print('Fail')
```

```python
#Range check 100 to 200
numbers = [150, 45, 200, 99, 120]
for num in numbers:
    if 100 <= num <= 200:
        print(num)
```

```python
#reverse loop with message
for i in range(10, 0, -1):
    if i == 5:
        print('Halfway there!')
    else:
        print(i)
```

```python
#sum of even numbers
total_sum = 0
for i in range(1, 11):
    if i % 2 == 0:
        total_sum += i
print('Sum of even numbers:', total_sum)
```

```python
#password match with break
passwords = ['123', 'abc', 'admin123', 'xyz']
for pwd in passwords:
    if pwd == 'admin123':
        print('Access Granted')
        break
```
